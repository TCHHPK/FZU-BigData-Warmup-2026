```python
import pandas as pd
import numpy as np
import os

DATA_DIR = "../data/processed/"

os.listdir(DATA_DIR)
```

检查文件齐全

```python
required_files = [
    "population_clean.csv",
    "age_clean.csv",
    "urban_clean.csv",
    "density_clean.csv",
    "wage_clean.csv",
    "unemployment_clean.csv",
    "employees_clean.csv",
    "industry_employment_clean.csv",
    "living_clean.csv"
]

for file in required_files:
    path = DATA_DIR + file

    print(
        file,
        "→",
        os.path.exists(path)
    )
```

读取全部已清洗数据

```python
population = pd.read_csv(
    DATA_DIR + "population_clean.csv"
)

age = pd.read_csv(
    DATA_DIR + "age_clean.csv"
)

urban = pd.read_csv(
    DATA_DIR + "urban_clean.csv"
)

density = pd.read_csv(
    DATA_DIR + "density_clean.csv"
)

wage = pd.read_csv(
    DATA_DIR + "wage_clean.csv"
)

unemployment = pd.read_csv(
    DATA_DIR + "unemployment_clean.csv"
)

employees = pd.read_csv(
    DATA_DIR + "employees_clean.csv"
)

industry = pd.read_csv(
    DATA_DIR + "industry_employment_clean.csv"
)

living = pd.read_csv(
    DATA_DIR + "living_clean.csv"
)
```

检查结构

```python
tables = {
    "population": population,
    "age": age,
    "urban": urban,
    "density": density,
    "wage": wage,
    "unemployment": unemployment,
    "employees": employees,
    "industry": industry,
    "living": living
}

for name, df in tables.items():

    print("=" * 60)
    print(name)

    print("Shape:", df.shape)
    print("Columns:", list(df.columns))

    print(
        "重复 city-year:",
        df.duplicated(["city_id", "year"]).sum()
    )
```

再次检查结构是否统一

```python
for name, df in tables.items():

    df["city_id"] = (
        df["city_id"]
        .astype(str)
        .str.strip()
    )

    df["year"] = pd.to_numeric(
        df["year"],
        errors="coerce"
    )
    
for name, df in tables.items():
    print(
        name,
        df["city_id"].dtype,
        df["year"].dtype
    )
```

合并+排序

```python
master = population.merge(
    age,
    on=["city_id", "year"],
    how="outer",
    validate="one_to_one"
)

master = master.merge(
    urban,
    on=["city_id", "year"],
    how="outer",
    validate="one_to_one"
)

master = master.merge(
    density,
    on=["city_id", "year"],
    how="outer",
    validate="one_to_one"
)


master = master.merge(
    wage,
    on=["city_id", "year"],
    how="outer",
    validate="one_to_one"
)

master = master.merge(
    unemployment,
    on=["city_id", "year"],
    how="outer",
    validate="one_to_one"
)

master = master.merge(
    employees,
    on=["city_id", "year"],
    how="outer",
    validate="one_to_one"
)

master = master.merge(
    industry,
    on=["city_id", "year"],
    how="outer",
    validate="one_to_one"
)

master = master.merge(
    living,
    on=["city_id", "year"],
    how="outer",
    validate="one_to_one"
)

master = (
    master
    .sort_values(["city_id", "year"])
    .reset_index(drop=True)
)

master.head(20)
```

计算每列缺失值，保存相关报告

```python
missing_report = pd.DataFrame({
    "missing_count": master.isna().sum(),
    "missing_rate": (
        master.isna().mean() * 100
    ).round(2)
})

missing_report.to_csv(
    DATA_DIR + "missing_report.csv"
)
```

保存总表

```python
master.to_csv(
    DATA_DIR + "master_data.csv",
    index=False
)

os.path.exists(
    DATA_DIR + "master_data.csv"
)
```
